```python
import subprocess
import sys
from pathlib import Path
HOST = '127.0.0.1'
PORT = 5555

DEVICE = 0
WIDTH = 960
HEIGHT = 540

NUM_HANDS = 2

MODEL_PATH = 'model/hand_landmarker.task'
CLASSIFIER_PATH = 'model/keypoint_classifier/keypoint_classifier.tflite'
LABELS_PATH = 'model/keypoint_classifier/keypoint_classifier_label.csv'

MIN_DETECTION_CONFIDENCE = 0.7
MIN_TRACKING_CONFIDENCE = 0.5

HEIGHT_DIFF_THRESHOLD = 0
MIN_GESTURE_CONFIDENCE = 0.6

RECONNECT_DELAY = 2.0

NO_DISPLAY = False
NO_FLIP = False

MAX_FRAMES = 0

DRY_RUN = False
VERBOSE = False

ARGS = [
    '--host', str(HOST),
    '--port', str(PORT),
    '--reconnect_delay', str(RECONNECT_DELAY),

    '--device', str(DEVICE),
    '--width', str(WIDTH),
    '--height', str(HEIGHT),

    '--model_path', MODEL_PATH,
    '--classifier_path', CLASSIFIER_PATH,
    '--labels_path', LABELS_PATH,

    '--num_hands', str(NUM_HANDS),

    '--min_detection_confidence', str(MIN_DETECTION_CONFIDENCE),
    '--min_tracking_confidence', str(MIN_TRACKING_CONFIDENCE),

    '--height_diff_threshold', str(HEIGHT_DIFF_THRESHOLD),
    '--min_gesture_confidence', str(MIN_GESTURE_CONFIDENCE),

    '--max_frames', str(MAX_FRAMES),
]

if NO_DISPLAY:
    ARGS.append('--no_display')

if NO_FLIP:
    ARGS.append('--no_flip')

if DRY_RUN:
    ARGS.append('--dry_run')

if VERBOSE:
    ARGS.append('--verbose')
```

```python
ROOT = str(Path.cwd())
proc = subprocess.run(
    [sys.executable, 'gesture_pipeline.py', *ARGS],
    cwd=ROOT
)

print('Exit code:', proc.returncode)
```
